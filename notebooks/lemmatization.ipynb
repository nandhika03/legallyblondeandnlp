### Lemmatization:
- It is similiar to stemming.
- But the result that we get after lemmatization is called a "lemma"
- It is called as a root word instead of a root stem.
- It gives proper valid meaningful words
- It takes longer to run than the Stemming
- It is a thin wrapper around the Word Net Corpus dictionary
- The class uses a morphy() to compare and find the lemma for each of the words.

```python
from nltk.stem import WordNetLemmatizer
lr = WordNetLemmatizer()
lr.lemmatize('silly',pos='a')
```

```python
from nltk.stem import WordNetLemmatizer
lr = WordNetLemmatizer()
lr.lemmatize('loving',pos='v')
```
