### Tokenization

- Tokenization is the process of breaking down text to its most nearest fundamental unit. 
- Paragraph -> sentences
- sentences -> words

- Punkt is NLTK’s pre-trained sentence boundary detection model.
- Punkt is the newer internal resource format used by recent NLTK versions.
- punkt → the actual sentence model
- punkt_tab → supporting tables / metadata that Punkt now depends on

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk.tokenize import sent_tokenize, word_tokenize, WordPunctTokenizer, TreebankWordTokenizer, RegexpTokenizer
elle = "C'mon  -- you have to help me pick out the perfect outfit. As they head toward the door, Elle spots AMY, a shy girl in a Delta Gamma Pledge sweatshirt, standing alone."
```

#### sent_tokenize
- This breaks down a string into a list of sentences.
- It handles punctuations 

```python
elle_sents = sent_tokenize(elle)
print(elle_sents)
```

#### word_tokenize
- This breaks down a sentence into a list of words and punctuations
- It handles contractions. 



```python
words_elle = word_tokenize(elle_sents[0])
print(words_elle)
```

#### WordPunctTokenizer
- This breaks down a sentence into a list of words and punctuations
- It also splits the contractions and punctuations

```python
wpt = WordPunctTokenizer()
wordpunct_elle = wpt.tokenize(elle_sents[0])
print(wordpunct_elle)
```

#### TreebankWordTokenizer
- This is commonly used for linguistic analysis (language analysis)
- It handles the english grammar in a better way and it handles contractions
- It splits punctuations

```python
tbwt = TreebankWordTokenizer()
tree_elle = tbwt.tokenize(elle_sents[0])
print(tree_elle)
```

#### RegexpTokenizer
- This splitts based on a pattern (regular expression)
- It handles punctuations 

```python
reg = RegexpTokenizer(r'\w+')

reg_elle = reg.tokenize(elle_sents[0])
print(reg_elle)
```
